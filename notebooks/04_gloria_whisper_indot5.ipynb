```python
from signbridge.audio.train_asr import compute_wer, compute_cer
from signbridge.text.pseudo_gloss import gloss_to_sentence_rules
from signbridge.bridge import BridgeModule

print('WER contoh:', compute_wer('saya mau tanya guru', 'saya ingin tanya guru'))
print('B0 ID:', gloss_to_sentence_rules('SAYA MAU TANYA GURU', lang='id'))
print('B0 FIL:', gloss_to_sentence_rules('SAYA MAU TANYA GURU', lang='fil'))

bridge = BridgeModule('data/bridge/concepts.csv')
print('Bridge ID -> FIL:', bridge.translate('Saya mau tanya guru tentang buku', src='id', tgt='fil'))
```
