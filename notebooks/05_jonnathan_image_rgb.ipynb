Membandingkan 3 keluarga arsitektur visi pada potongan tangan `224x224`:
- **I1**: `MobileNetV3SmallClassifier` (CNN terpisahkan kedalaman, kandidat browser).
- **I2**: `EfficientNetB0Classifier` (CNN skala majemuk).
- **I3**: `MobileViTXSClassifier` (Hibrida CNN + Transformer Self-Attention).

```python
import torch
from signbridge.image.train_image import build_image_model, export_fingerspell_onnx

x = torch.randn(2, 3, 224, 224)
for arch in ['mobilenetv3_small', 'efficientnet_b0', 'mobilevit_xs']:
    m = build_image_model(arch=arch, num_classes=26)
    out = m(x)
    params = sum(p.numel() for p in m.parameters()) / 1e6
    print(f'{arch:20s} -> output {tuple(out.shape)} | {params:.3f}M params')
```
